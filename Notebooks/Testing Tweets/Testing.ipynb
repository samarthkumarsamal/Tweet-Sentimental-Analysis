#### Importing Python Libraries

```python
import numpy as np
import pandas as pd
import joblib
```

#### Loading the Models

```python
models = {
    "LR_TFIDF": joblib.load("../../Models/lor-tfidf.joblib"),
    "LR_COUNT": joblib.load("../../Models/lor-count.joblib"),
    "NB_TFIDF": joblib.load("../../Models/nb-tfidf.joblib"),
    "NB_COUNT": joblib.load("../../Models/nb-count.joblib"),
}

print("All Models Loaded Successfully")
```

#### User Input

```python
def get_user_tweet():
    tweet = input("\nEnter a Tweet (0 to exit): ").strip()
    return tweet if tweet else " "
```

#### Prediction Function

```python
def prediction(model, tweet):
    tweet_series = pd.Series([tweet]).astype(str)

    predicted_label = int(model.predict(tweet_series)[0])
    prob_neg, prob_pos = model.predict_proba(tweet_series)[0]

    return {
        "Sentiment": "Positive (1)" if predicted_label == 1 else "Negative (0)",
        "Predicted_Label": predicted_label,
        "Prob_Negative": prob_neg,
        "Prob_Positive": prob_pos,
        "Confidence": prob_pos if predicted_label == 1 else prob_neg
    }
```

#### Running Predictions

```python
results_history = []

while True:
    tweet = get_user_tweet()
    if tweet == "0":
        break
    tweet_results = {"Tweet": tweet}
    for model_name, model in models.items():
        tweet_results[model_name] = prediction(model, tweet)
    results_history.append(tweet_results)


print("\nFINAL COMPARATIVE SUMMARY")
print("=" * 90)

for idx, entry in enumerate(results_history, start=1):
    print(f"\nTweet {idx}: {entry['Tweet']}")
    print("-" * 90)

    table_data = []

    for model_name in models.keys():
        res = entry[model_name]
        table_data.append([
            model_name,
            res["Sentiment"],
            f"{res['Confidence']:.4f}",
            f"{res['Prob_Negative']:.4f}",
            f"{res['Prob_Positive']:.4f}",
        ])

    df = pd.DataFrame(
        table_data,
        columns=[
            "Model",
            "Sentiment",
            "Confidence",
            "Prob Negative (0)",
            "Prob Positive (1)"
        ]
    )

    print(df.to_string(index=False))
    print("=" * 50)
```
